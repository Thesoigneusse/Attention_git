```python
import os

heads = [ 186, 335, 457, 690, 792, 1034, 1268, 1420, 1610, 1755, 1782, 1814, 1863, 1910, 1932, 1958, 1989, 2009, 2044, 2075, 2108, 2123, 2153, 2191, 2214, 2245, 2262, 2280]


def lister_fichiers_pour_combinaisons(_id_min, _id_max):
    for nombre1 in range(6 if _concat else 8):
        for nombre2 in range(8):
            nom_fichier_sortie = f"{_OUTPUT_PATH}/list_chemin_{nombre1}_{nombre2}{suffixe}.txt"
            
            # Liste pour stocker les chemins de fichiers trouvés
            fichiers = []
            offset = 0
            # Parcours les dossiers {_id} de _id_min à _id_max
            for _id in range(_id_min, _id_max + 1):
                identifiant = _id + offset
                if _id not in heads and _id -1 not in heads:
                    # Construit le chemin du dossier cible
                    chemin_dossier = os.path.join(_INPUT_PATH, str(identifiant), "crt_to_full_ctxs_crt", str(nombre1)) #, str(nombre2))
                    
                    # Ajoute les fichiers trouvés à la liste principale
                    fichiers.append(f"{chemin_dossier}/{nombre2}.tsv")
                elif _id in heads:
                    offset += 1
            
            # Écrit les chemins des fichiers trouvés dans le fichier de sortie
            try:
                with open(nom_fichier_sortie, "w") as f_out:
                    for fichier in fichiers:
                        f_out.write(f"{fichier}\n")
                
                print(f"Fichier '{nom_fichier_sortie}' créé avec {len(fichiers)} fichiers.")
            
            except Exception as e:
                print(f"Erreur lors de l'écriture du fichier '{nom_fichier_sortie}': {e}")


_id_min = 2
_id_max = 2278

lister_fichiers_pour_combinaisons(_id_min, _id_max)
```
